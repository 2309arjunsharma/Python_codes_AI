```python
num  = int(input("Enter a number: "))
if num%2==0:
    print("The number", num, "is even.")
else:
    print("The number", num, "is odd.")
```

```python
percent = float(input("Enter the percentage: "))
att = float(input("Enter the attendance percentage: "))

if percent >= 45 and att >= 75:
    print("Promoted")
else:
    print("Not Promoted")
```

```python
temp = float(input("Enter the temperature: "))
if temp>40 or temp<0:
    print("Extreme temperature")
else:
    print("Normal temperature")
```

```python
cart_value = float(input("Enter the cart value: "))
dist = float(input("Enter the distance: "))
if cart_value > 1000 and dist < 10:
    print("Free delivery")
else:
    charge = dist * 5  
    print("Delivery charges apply", charge)
```

```python
net_income = float(input("Enter the net income: ")) 
print("The net income is", net_income)
if net_income <= 250000:
    print("No tax")
else:
    print("Tax is 5%")
    tax = net_income * 0.05
    print("Tax amount is", tax)
    net_income_after_tax = net_income - tax
    print("Net income after tax is", net_income_after_tax)
```
